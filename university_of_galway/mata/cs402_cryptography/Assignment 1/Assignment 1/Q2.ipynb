```python
import cs402
from cs402 import *
# Locate the file of ciphertext downloaded from assignment
file = open("C:\\Users\\cianl\\Downloads\\CS402\\Assignment 1\\a1q2-cipher6.txt") 

# Read the contents of the text file
CipherText = file.read()

# Use the string_to_int_list function from 'cs402.py' to convert the CipherText into
# the number each letter represents in the 68-symbol alphabet

CipherInt = string_to_int_list(ALPHABET68,CipherText)

# declare a shift cipher using the 68 symbol alphabet
Q2=ShiftCipher(ALPHABET68)

# for-loop to cycle through all possible keys 'a'

for a in range(1,67):
    
    # ShiftCipher.decrypt_int_list() function shifts the CipherInt values back to its original using key
    
    DecipheredInt = ShiftCipher.decrypt_int_list(Q2,a,CipherInt)
    
    # int_list_to_string() converts the new integer values back to string text
    DecipheredText = int_list_to_string(ALPHABET68,DecipheredInt)
    
    # print each Key and DecipheredText to see which key decrypted the CipherText properly
    print("Key", a)
    print(DecipheredText)
    

```
