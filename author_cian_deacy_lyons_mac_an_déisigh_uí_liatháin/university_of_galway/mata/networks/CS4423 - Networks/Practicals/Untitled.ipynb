```python
import networkx as nx
opts = { "with_labels": True, "node_color": 'y', "font_color" : 'r' }
```

```python
G = nx.Graph(["12", "13", "24","35","36","45","46","57","67","68", "78"])
nx.draw(G,**opts)
```

```python
G.nodes()
```

```python
N = nx.neighbors(G, '2') 
S = G.subgraph(list(N)) 
nx.draw(S, **opts)
```

```python
def circle_graph(n, d): 
    G = nx.cycle_graph(n) 
    for v in G:
        for o in range(2, d+1): 
            G.add_edge(v, (v+o) % n)
    return G
```

```python
G = circle_graph(6,2)
```

```python
nx.draw_circular(G,**opts)
```
